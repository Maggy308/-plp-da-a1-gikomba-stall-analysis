# Gikomba Stall Analysis
## PLP Data Analysis Assignment 1

**Name:** Magdaline Wanjiku Kibui

**Cohort:** 7

This project analyses 14 days of sales and stock data for Wanjiku's fresh produce stall at Gikomba Market. The objective is to understand sales performance, stock requirements and product profitability using core Python.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```


## Task 1: Daily Sales Analysis

In this task, I calculate total sales, average daily sales, the highest and lowest sales, and identify the day and week when the highest and lowest sales occurred.

```python
# Calculate total sales
total_sales = sum(daily_sales)

# Calculate average daily sales
avg_sale = total_sales / len(daily_sales)

# Identify highest and lowest sales
highest_sale = max(daily_sales)
lowest_sale = min(daily_sales)

# Identify best and worst days
best_index = daily_sales.index(highest_sale)
worst_index = daily_sales.index(lowest_sale)

best_day = days[best_index]
worst_day = days[worst_index]

# Identify the corresponding weeks
best_week = best_index // 7 + 1
worst_week = worst_index // 7 + 1

print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {avg_sale:,.2f}")
print(f"Highest Daily Sales: KES {highest_sale:,.2f}")
print(f"Lowest Daily Sales: KES {lowest_sale:,.2f}")
print(f"Best Day: {best_day}, Week {best_week}")
print(f"Worst Day: {worst_day}, Week {worst_week}")
```

## Task 2: Weekly Sales Analysis

In this task, I use a function and list slicing to calculate total sales for each week. I then compare the two weeks to determine the percentage growth or decline in sales.

```python
def week_totals(sales):
    """Return the total sales for week one and week two."""

    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])

    return week1, week2


# Call the function and unpack its results
week1, week2 = week_totals(daily_sales)

# Calculate week-on-week growth
growth_percentage = ((week2 - week1) / week1) * 100

print(f"Week 1 Total: KES {week1:,.2f}")
print(f"Week 2 Total: KES {week2:,.2f}")
print(f"Week-on-Week Growth: {growth_percentage:.2f}%")
```

## Task 3: Classifying Daily Sales

In this task, I create a function to classify daily sales as Strong, Normal or Weak based on the average daily sales. I use a dictionary to count how many days fall into each category.

```python
def classify_day(amount, average):
    """Classify daily sales as Strong, Normal or Weak."""

    if amount > average * 1.20:
        return "Strong"

    elif amount < average * 0.80:
        return "Weak"

    else:
        return "Normal"
counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

for amount in daily_sales:
    category = classify_day(amount, avg_sale)
    counts[category] += 1

print(counts)

```

## Task 4: Weekday Versus Weekend Sales

In this task, I calculate average weekday sales from Monday to Friday and average weekend sales from Saturday to Sunday across both weeks. I then calculate the percentage difference using weekday sales as the baseline.

```python
def avg_weekdays(sales):
    """Calculate average sales for weekdays across two weeks."""

    weekdays = sales[:5] + sales[7:12]

    return sum(weekdays) / len(weekdays)


def avg_weekends(sales):
    """Calculate average sales for weekends across two weeks."""

    weekends = sales[5:7] + sales[12:14]

    return sum(weekends) / len(weekends)


weekday_avg = avg_weekdays(daily_sales)
weekend_avg = avg_weekends(daily_sales)

difference = weekend_avg - weekday_avg

percentage_difference = (difference / weekday_avg) * 100

print(f"Average Weekday Sales: KES {weekday_avg:,.2f}")
print(f"Average Weekend Sales: KES {weekend_avg:,.2f}")
print(f"Sales Difference: KES {difference:,.2f}")
print(f"Percentage Difference: {percentage_difference:.2f}%")
```

## Task 5: Stock Management and Gross Margins

In this task, I use the stock dictionary to identify products whose quantities are at or below their reorder levels. I also calculate the gross margin percentage for every product using its cost and selling price.

```python
def below_reorder(stock):
    """Return a list of products at or below reorder level."""

    products = []

    for item, details in stock.items():
        if details["qty"] <= details["reorder"]:
            products.append(item)

    return products


def gross_margin(stock):
    """Calculate the gross margin percentage for each product."""

    margins = {}

    for item, details in stock.items():
        margin = (
            (details["price"] - details["cost"])
            / details["price"]
        ) * 100

        margins[item] = margin

    return margins


# Display products requiring restocking
print("Products Requiring Restocking:")

for product in below_reorder(stock):
    print(product)


# Display gross margins
print("\nGross Margin by Product:")

for product, margin in gross_margin(stock).items():
    print(f"{product}: {margin:.2f}%")


```

## Task 6: Stall Performance Report

```python
print("=" * 55)
print("              STALL PERFORMANCE REPORT")
print("=" * 55)

print("\nSALES PERFORMANCE")
print("-" * 55)

print(f"{'Total Sales:':<30} KES {total_sales:>12,.2f}")
print(f"{'Average Daily Sales:':<30} KES {avg_sale:>12,.2f}")
print(f"{'Highest Daily Sales:':<30} KES {highest_sale:>12,.2f}")
print(f"{'Lowest Daily Sales:':<30} KES {lowest_sale:>12,.2f}")

print(f"{'Best Sales Day:':<30} {best_day}, Week {best_week}")
print(f"{'Worst Sales Day:':<30} {worst_day}, Week {worst_week}")

print("\nWEEKLY PERFORMANCE")
print("-" * 55)

print(f"{'Week 1 Sales:':<30} KES {week1:>12,.2f}")
print(f"{'Week 2 Sales:':<30} KES {week2:>12,.2f}")
print(f"{'Weekly Growth:':<30} {growth_percentage:>12.2f}%")

print("\nWEEKDAY AND WEEKEND PERFORMANCE")
print("-" * 55)

print(f"{'Average Weekday Sales:':<30} KES {weekday_avg:>12,.2f}")
print(f"{'Average Weekend Sales:':<30} KES {weekend_avg:>12,.2f}")
print(f"{'Weekend Sales Advantage:':<30} {percentage_difference:>12.2f}%")

print("\nDAILY SALES CLASSIFICATION")
print("-" * 55)

for category, count in counts.items():
    print(f"{category:<30} {count:>12} days")

print("\nSTOCK MANAGEMENT")
print("-" * 55)

print("Products Requiring Restocking:")

for product in below_reorder(stock):
    print(f"- {product}")

print("\nGROSS MARGINS")
print("-" * 55)

for product, margin in gross_margin(stock).items():
    print(f"{product:<30} {margin:>11.2f}%")

print("=" * 55)
print("             END OF PERFORMANCE REPORT")
print("=" * 55)

```
