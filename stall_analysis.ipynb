```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700] # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2



# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9,"reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":{"qty": 12, "reorder": 10, "cost": 80, "price": 140},
    }
```

Task 1

```python
#total sales for the two weeks
total_sales=sum(daily_sales)
print(f"Total sales: {total_sales}")

#Average sales for the 2 weeks
avg_sale=sum(daily_sales)/len(daily_sales)
print(f"Average sale: {avg_sale}")

#highest daily sales
highest_sale=max(daily_sales)
print(f" Highest sale: {highest_sale}")

#lowest daily sales
lowest_sale=min(daily_sales)
print(f"Lowest sale: {lowest_sale}")

#best day
best_day=days[daily_sales.index(max(daily_sales))]
print(f" Best day: {best_day}")

#worst day
worst_day=days[daily_sales.index(min(daily_sales))]
print(f"Worst day:{worst_day}")

best_index = daily_sales.index(highest_sale)
worst_index = daily_sales.index(lowest_sale)

best_week = best_index // 7 + 1
worst_week = worst_index // 7 + 1

print(f"Best day: {days[best_index]}, Week {best_week}")
print(f"Worst day: {days[worst_index]}, Week {worst_week}")
```

Task 2

```python
def week_totals(sales):
    """Return total sales for week one and week two."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])

    return week1, week2

    week1, week2 = week_totals(daily_sales)

growth_percentage = ((week2 - week1) / week1) * 100

print(f"Week 1: KES {week1:,}")
print(f"Week 2: KES {week2:,}")
print(f"Growth: {growth_percentage:.2f}%")
```

Task 3

```python
def classify_day(amount, average):
    """Classify daily sales as Strong, Normal or Weak."""

    if amount > average * 1.20:
        return "Strong"

    elif amount < average * 0.80:
        return "Weak"

    else:
        return "Normal"
counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

for amount in daily_sales:
    category = classify_day(amount, avg_sale)
    counts[category] += 1

print(counts)

```

Task 4

```python
def avg_weekdays(sales):
    """Calculate average sales from Monday to Friday."""
    weekdays = sales[:5] + sales[7:12]
    return sum(weekdays) / len(weekdays)


def avg_weekends(sales):
    """Calculate average sales for Saturday and Sunday."""
    weekends = sales[5:7] + sales[12:14]
    return sum(weekends) / len(weekends)


weekday_avg = avg_weekdays(daily_sales)
weekend_avg = avg_weekends(daily_sales)

difference = weekend_avg - weekday_avg

percentage_difference = (
    difference / weekday_avg
) * 100

print(f"Weekday average: KES {weekday_avg:,.2f}")
print(f"Weekend average: KES {weekend_avg:,.2f}")
print(f"Difference: {percentage_difference:.2f}%")

```

Task 5

```python
stock = {

"Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

"Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},

"Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},

"Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},

"Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
def below_reorder(stock):
    """Return products at or below their reorder levels."""

    products = []

    for item, details in stock.items():
        if details["qty"] <= details["reorder"]:
            products.append(item)

    return products


print(below_reorder(stock))

def gross_margin(stock):
    """Calculate gross margin percentage for every product."""

    margins = {}

    for item, details in stock.items():
        margin = (
            (details["price"] - details["cost"])
            / details["price"]
        ) * 100

        margins[item] = margin

    return margins


print(gross_margin(stock))


```
